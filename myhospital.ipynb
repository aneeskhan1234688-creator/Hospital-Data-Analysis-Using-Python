```python
# How many hospitals are in each state
data["State"].value_counts()
```

```python
# Fill text columns that are missing a lot of values
text_cols_to_fill = ["Website", "Description", "Local Hospital Network (LHN)", "Primary Health Network area (PHN)"]
data[text_cols_to_fill] = data[text_cols_to_fill].fillna("Not Available")


# Drop the few rows that are missing contact or location info
data = data.dropna(subset=["Phone number", "Street address", "Suburb"])

# Check null values again after cleaning
data.isnull().sum()
```

```python
data["State"]
```

```python
data["Beds_Number"] = data["Beds"].map({
    "<50": 25,
    "50-99": 75,
    "100-199": 150,
    "200-500": 350,
    ">500": 500
})
```

```python
data["Sector"] = data["Sector"].fillna("Private")
```

```python
data["Sector"].value_counts()
```

```python
data.isnull().sum()
```

```python
# Fill missing numerical bed values with the median
data["Beds_Number"] = data["Beds_Number"].fillna(data["Beds_Number"].median())

print(data[["Beds", "Beds_Number"]].head(10))
```

```python
# Hospitals in New South Wales
nsw_hospitals = data[data["State"] == "NSW"]

nsw_hospitals[["Hospital name", "State", "Sector", "Beds"]].head()
```

```python
# Hospitals with more than 200 beds
large_hospitals = data[data["Beds_Number"] > 200]

large_hospitals[["Hospital name", "State", "Sector", "Beds", "Beds_Number"]].head()
```

```python
# Public hospitals with more than 200 beds
large_public_hospitals = data[(data["Sector"] == "Public") & (data["Beds_Number"] > 200)]

large_public_hospitals[["Hospital name", "State", "Sector", "Beds"]].head()
```

```python
# Select important columns
hospital_details = data[
    ["Hospital name", "State", "Sector", "Beds", "Beds_Number"]
]

hospital_details.head()
```

```python
# Sort hospitals by number of beds
sorted_hospitals = data.sort_values("Beds_Number", ascending=False)

sorted_hospitals[["Hospital name", "State", "Sector", "Beds", "Beds_Number"]].head(10)
```

```python
# Number of hospitals in each bed category
bed_count = data["Beds"].value_counts()

print(bed_count)
```

```python
# Hospital count by State
state_group = data.groupby("State")["Hospital_ID"].count()

print(state_group)
```

```python
# Hospital count by Sector
sector_group = data.groupby("Sector")["Hospital_ID"].count()

print(sector_group)
```

```python
# Average numerical bed value by Sector
average_beds_sector = data.groupby("Sector")["Beds_Number"].mean()

print(average_beds_sector)
```

```python
# Chart 1: Hospitals by State
state_count = data["State"].value_counts()
plt.figure(figsize=(15,6))
sns.barplot(x=state_count.index, y=state_count.values, color="orange")
plt.title("Hospitals by State")
plt.xlabel("State")
plt.ylabel("Number of Hospitals")
plt.show()
```

```python
data.columns
```

```python
# # Chart 2: Top 5 Local Hospital Networks by Number of Hospitals
lhn_count = data["Local Hospital Network (LHN)"].value_counts().head(5)

plt.figure(figsize=(10, 6))

plt.pie(
    lhn_count.values,
    labels=lhn_count.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Hospital Distribution by Top 5 Local Hospital Networks")
plt.show()
```

```python
# CHART 3 — Distribution of hospital bed values
plt.figure(figsize=(15, 6))
sns.histplot(data=data, x="Beds_Number", bins=10, kde=True, color="green", label="Beds_Number")
plt.title("Distribution of Hospital Bed Values")
plt.xlabel("Number of Beds")
plt.ylabel("Number of Patients")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# Chart 4: Bed Distribution
plt.figure(figsize=(18,7))
sns.histplot(data["Hospital_ID"], bins=30, edgecolor = 'white', color="red", alpha=0.5, label="Hospital_ID")
plt.title("Distribution of Hospital Beds")
plt.xlabel("Number of Beds")
plt.ylabel("Number of Hospitals")
plt.legend()
plt.show()
```

```python

# CHART 5 :  Geographic distribution of hospitals

plt.figure(figsize=(9, 6))
sns.scatterplot(
    data=data,
    x="Longitude",
    y="Latitude",
    hue="Sector", 
    alpha=0.65
)
plt.title("Geographic Distribution of Hospitals")
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.legend(title="Sector")
plt.tight_layout()
plt.show()
```

```python
# Chart 6: Top 10 Suburbs with Hospitals
top_suburbs = data["Suburb"].value_counts().head(10)
plt.figure(figsize=(15,6))
sns.barplot(x=top_suburbs.values, y=top_suburbs.index, color="teal", label="Number of Hospital")
plt.title("Top 10 Suburbs by Hospital Count")
plt.xlabel("Number of Hospitals")
plt.ylabel("Suburb")
plt.tight_layout()
plt.legend()
plt.show()

```

```python
# Chart 7: Average Beds by State
avg_beds = data.groupby("State")["Beds_Number"].mean()
plt.figure(figsize=(9,5))
sns.lineplot(x=avg_beds.index, y=avg_beds.values, color="purple", marker="s", label="Average Beds")
plt.title("Average Hospital Beds by State")
plt.xlabel("State")
plt.ylabel("Average Beds")
plt.tight_layout()
plt.legend()
plt.show()
```

```python
# Chart 8: Hospitals by Bed Category
bed_count = data["Beds"].value_counts()
plt.figure(figsize=(8,5))
sns.barplot(x=bed_count.index, y=bed_count.values, color="coral", edgecolor="black", label="Bed Category Count")
plt.title("Hospitals by Bed Category")
plt.xlabel("Bed Category")
plt.ylabel("Hospital Count")
plt.legend()
plt.show()
```

```python
# Chart 9: Bed Distribution by Hospital Sector

plt.figure(figsize=(7,5))
sns.boxplot(data=data, x="Sector", y="Beds_Number", label="Sector")
plt.title("Bed Distribution by Sector")
plt.xlabel("Sector")
plt.ylabel("Number of Beds")
plt.legend()
plt.show()
```

```python
# Chart 10: pairplot
plt.figure(figsize=(10,7))
sns.pairplot(data[["Beds_Number", "Latitude", "Longitude", "Sector"]], hue="Sector")
```

```python
# Chart 11: Bed Distribution by Sector

plt.figure(figsize=(8, 5))
sns.violinplot(data=data, x="Sector", y="Beds_Number", color="deeppink", alpha=0.6)

plt.title("Bed Distribution by Sector")
plt.xlabel("Sector")
plt.ylabel("Number of Beds")
plt.show()
```

```python
# Chart 12: Hospital Bed Category by Sector
plt.figure(figsize=(17,6))
sns.countplot(data=data, x="Beds", hue="Sector", alpha=0.7)

plt.title("Bed Categories by Sector")
plt.xlabel("Bed Category")
plt.ylabel("Number of Hospitals")
plt.legend()
plt.show()
```

```python
# Chart 13: Density Distribution by Hospital Beds

plt.figure(figsize=(17, 6))

sns.kdeplot(
    data=data,
    x="Beds_Number",
    color="purple",
    linewidth=5,
    alpha=0.4,
    fill=True,
    label="Number of Beds"
)

plt.title("Density Distribution of Hospital Beds")
plt.xlabel("Number of Beds")
plt.ylabel("Density")
plt.legend()

plt.tight_layout()
plt.show()
```

```python
# Chart 14: Heatmap

sns.heatmap(data.corr(numeric_only=True), annot=True, cmap="coolwarm")
```

```python
# Dashboard 1 - Hospital Overview

fig, axes = plt.subplots(2, 2, figsize=(15, 13))

# 1. Bar Chart
state_count = data["State"].value_counts()

axes[0,0].bar(state_count.index, state_count.values, color="pink")
axes[0,0].set_title("Hospitals by State")
axes[0,0].set_xlabel("State")
axes[0,0].set_ylabel("Number of Hospitals")
axes[0,0].tick_params(axis="x", rotation=45)


# 2. Pie Chart
lhn_count = data["Local Hospital Network (LHN)"].value_counts().head(5)

axes[0,1].pie(
    lhn_count.values,
    labels=lhn_count.index,
    autopct="%1.1f%%"
)

axes[0,1].set_title("Local Hospital Network")

# 3. Line Chart
state_beds = data.groupby("State")["Hospital_ID"].count()

sns.lineplot(x=state_beds.index,
             y=state_beds.values,
             marker="o",
             color="purple",
             ax=axes[1,0])

axes[1,0].set_title("Hospital Count by State")
axes[1,0].set_xlabel("State")
axes[1,0].set_ylabel("Hospitals")
axes[1,0].tick_params(axis="x", rotation=45)


# 4. Countplot
sns.countplot(data=data,
              x="Beds",
              ax=axes[1,1], color="cyan")

axes[1,1].set_title("Hospitals by Bed Category")
axes[1,1].set_xlabel("Bed Category")
axes[1,1].set_ylabel("Count")
```

```python
# Dashboard 2 - Hospital Capacity

fig, axes = plt.subplots(2, 2, figsize=(16, 10))

# 5. Histogram
sns.histplot(data=data,
             x="Beds_Number",
             bins=10,
             kde=True,
             ax=axes[0,0],
             color="orange",
             label="Number of Beds"
             )
axes[0,0].set_title("Distribution of Beds")
axes[0,0].set_xlabel("Number of Beds")
axes[0,0].set_ylabel("Count")
axes[0,0].legend()


# 6. Boxplot
sns.boxplot(data=data,
            x="Sector",
            y="Beds_Number",
            ax=axes[0,1],
            color="coral",
            label="Beds by Sector"
            )
axes[0,1].set_title("Beds by Sector")
axes[0,1].set_xlabel("Sector")
axes[0,1].set_ylabel("Number of Beds")
axes[0,1].legend()


# 7. Violin Plot
sns.violinplot(data=data,
               x="Sector",
               y="Beds_Number",
               ax=axes[1,0],
               color="teal",
               alpha=0.6,
               label="Bed Distribution"
               )
axes[1,0].set_title("Bed Distribution by Sector")
axes[1,0].set_xlabel("Sector")
axes[1,0].set_ylabel("Number of Beds")
axes[1,0].legend()

#8

sns.histplot(data["Hospital_ID"],
        bins=15,
        edgecolor="white",
        color="red",
        alpha=0.5,
        ax=axes[1,1],
        label="Hospital ID Count"
)

axes[1,1].set_title("Distribution of Hospital Beds")
axes[1,1].set_xlabel("Number of Beds")
axes[1,1].set_ylabel("Number of Hospitals")
axes[1,1].legend()
plt.show()
```

```python
# Dashboard 3 - Hospital Location & Distribution

fig, axes = plt.subplots(3, 2, figsize=(15, 15))


# 9. Scatter Plot
sns.scatterplot(data=data,
                x="Longitude",
                y="Latitude",
                hue="Sector",
                ax=axes[0,0])

axes[0,0].set_title("Hospital Locations")
axes[0,0].set_xlabel("Longitude")
axes[0,0].set_ylabel("Latitude")


# 10. Heatmap
state_sector = pd.crosstab(data["State"], data["Sector"])

sns.heatmap(state_sector,
            annot=True,
            fmt="d",
            cmap="Blues",
            ax=axes[0,1])

axes[0,1].set_title("State vs Sector")
axes[0,1].set_xlabel("Sector")
axes[0,1].set_ylabel("State")


# 11. KDE Plot
sns.kdeplot(data=data,
            x="Beds_Number",
            fill=True,
            color="purple",
            ax=axes[1,0])

axes[1,0].set_title("Bed Density")
axes[1,0].set_xlabel("Number of Beds")
axes[1,0].set_ylabel("Density")


# 12. Horizontal Bar
suburb_count = data["Suburb"].value_counts().head(10)

sns.barplot(x=suburb_count.values,
            y=suburb_count.index,
            color="coral",
            ax=axes[1,1])

axes[1,1].set_title("Top 10 Suburbs")
axes[1,1].set_xlabel("Number of Hospitals")
axes[1,1].set_ylabel("Suburb")


# 13. Countplot
sns.countplot(data=data,
              x="Sector",
              hue="Beds",
              ax=axes[2,0])

axes[2,0].set_title("Sector vs Bed Category")
axes[2,0].set_xlabel("Sector")
axes[2,0].set_ylabel("Count")


# 14. Histogram
sns.histplot(data=data,
             x="Beds_Number",
             bins=10,
             color="gold",
             ax=axes[2,1])

axes[2,1].set_title("Distribution of Beds")
axes[2,1].set_xlabel("Number of Beds")
axes[2,1].set_ylabel("Count")


plt.tight_layout()
plt.show()
```
